```python
#멀티플롯 
import matplotlib.pyplot as plt

# 시계열 데이터 (t, TAST, Voltage, Current)
data = [
    (0.0, 0, 0, 0),
    (0.2, 0, 0, 0),
    (0.4, 1606, 80, 62),
    (0.6, 1620, 80, 62),
    (0.8, 2196, 244, 110),
    (1.0, 3277, 227, 138),
    (1.2, 1766, 278, 134),
    (1.4, 3729, 233, 65),
    (1.7, 3457, 233, 65),
    (1.9, 3520, 148, 97),
    (2.1, 3728, 194, 52),
    (2.3, 3606, 151, 90),
    (2.5, 3858, 178, 65),
    (2.7, 3654, 145, 94),
    (2.9, 3504, 176, 66),
    (3.1, 3879, 150, 93),
]

# 리스트 언팩
xs, tast, volt, curr = zip(*data)

# 스타일 설정
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = [10, 8]
plt.rcParams['font.size'] = 10

# 서브플롯 생성
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, sharex=True, gridspec_kw={'hspace': 0.3})

# 1. TAST
ax1.plot(xs, tast, 'r-', linewidth=1.5, label='TAST')
ax1.set_ylabel('TAST', fontweight='bold')
ax1.grid(True, linestyle='--', alpha=0.7)
ax1.legend(loc='upper right')

# 2. Voltage (0.1V → V)
ax2.plot(xs, volt, 'g-', linewidth=1.5, label='Voltage')
ax2.set_ylabel('Voltage (V)', fontweight='bold')
ax2.grid(True, linestyle='--', alpha=0.7)
ax2.legend(loc='upper right')

# 3. Current (0.1A → A)
ax3.plot(xs, curr, 'b-', linewidth=1.5, label='Current')
ax3.set_xlabel('Time (s)', fontweight='bold')
ax3.set_ylabel('Current (A)', fontweight='bold')
ax3.grid(True, linestyle='--', alpha=0.7)
ax3.legend(loc='upper right')

# 타이틀 및 레이아웃 조정
plt.suptitle('Kemppi Welding Parameters', fontsize=14, y=0.98)
plt.tight_layout()
plt.show()
```

```python
#상관관계 분석 
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 시계열 데이터 (t, TAST, Voltage, Current)
data = [
    (0.0, 0, 0, 0),
    (0.2, 0, 0, 0),
    (0.4, 1606, 80, 62),
    (0.6, 1620, 80, 62),
    (0.8, 2196, 244, 110),
    (1.0, 3277, 227, 138),
    (1.2, 1766, 278, 134),
    (1.4, 3729, 233, 65),
    (1.7, 3457, 233, 65),
    (1.9, 3520, 148, 97),
    (2.1, 3728, 194, 52),
    (2.3, 3606, 151, 90),
    (2.5, 3858, 178, 65),
    (2.7, 3654, 145, 94),
    (2.9, 3504, 176, 66),
    (3.1, 3879, 150, 93),
]

# 데이터프레임으로 변환
df = pd.DataFrame(data, columns=['Time', 'TAST', 'Voltage', 'Current'])

# 상관계수 행렬 계산
corr = df.corr()

# 히트맵 스타일 설정
plt.figure(figsize=(8, 6))
sns.set(font_scale=1.1)
sns.set_style("white")

# 히트맵 그리기
mask = np.triu(np.ones_like(corr, dtype=bool))  # 상삼각형 마스크
heatmap = sns.heatmap(corr, 
                     mask=mask,
                     annot=True, 
                     fmt=".2f", 
                     cmap='coolwarm',
                     vmin=-1, 
                     vmax=1,
                     linewidths=0.5,
                     square=True,
                     cbar_kws={"shrink": 0.8})

# 제목 및 레이아웃 조정
plt.title('상관관계 분석 히트맵', pad=15, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# 상관계수 행렬 출력
print("\n상관계수 행렬:")
print(corr.round(2))
```
